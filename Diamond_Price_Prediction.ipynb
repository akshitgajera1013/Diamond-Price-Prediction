```python
num_col=df.select_dtypes(include='number')
```

```python
from numpy._core.defchararray import lower
q1=num_col.quantile(0.25)
q3=num_col.quantile(0.75)
iqr=q3-q1

lower_bound=q1-1.5*iqr
upper_bound=q3+1.5*iqr

outlier_col=(num_col<lower_bound)|(num_col>upper_bound)
print(outlier_col.sum())
```

```python
# Remove rows that contain at least one outlier
df = df[~outlier_col.any(axis=1)]

print(f"New shape of the dataset after removing outliers: {df.shape}")
display(df.head())
```

```python
from scipy import stats

# Define categorical columns and target
categorical_cols = ['Cut(Quality)', 'Color', 'Clarity']
target = 'Price(in US dollars)'

anova_results = {}

for col in categorical_cols:
    # Group the price data by each category in the column
    groups = [group[target].values for name, group in df.groupby(col)]

    # Perform One-way ANOVA
    f_stat, p_val = stats.f_oneway(*groups)
    anova_results[col] = {'F-Statistic': f_stat, 'p-value': p_val}

# Convert results to DataFrame for display
anova_df = pd.DataFrame(anova_results).T
display(anova_df)

# Interpretation
for col, results in anova_results.items():
    if results['p-value'] < 0.05:
        print(f"{col} IS statistically significant (important) for predicting price (p < 0.05).")
    else:
        print(f"{col} is NOT statistically significant for predicting price (p >= 0.05).")
```

```python
x=df.drop('Price(in US dollars)',axis=1)
y=df['Price(in US dollars)']
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler,OneHotEncoder,FunctionTransformer,OrdinalEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.metrics import r2_score,mean_absolute_error
```

```python
num_col.skew()
```

```python
skwed_cols=['Carat(Weight of Daimond)','Depth','Y(width)','Z(Depth)']
```

```python
ordinal_col=['Cut(Quality)']
```

```python
df.head()
```

```python
one_hot_cols=['Color','Clarity']
```

```python
skew_pipeline=Pipeline(steps=[
    ('skew',FunctionTransformer(np.log1p)),
    ('scale',StandardScaler())
])

num_pipeline=Pipeline(steps=[
    ('scale',StandardScaler())
])

ordinal_pipeline=Pipeline(steps=[
    ('encoder',OrdinalEncoder(categories=[['Fair','Good','Very Good','Premium','Ideal']]))
])

cat_pipeline=Pipeline(steps=[
    ('one-hot',OneHotEncoder())
])
```

```python
skwed_cols
```

```python
num_col.head()
```

```python
preprocessor=ColumnTransformer(transformers=[
    ('skew',skew_pipeline,skwed_cols),
    ('num',num_pipeline,['Table','X(length)']),
    ('ordinal',ordinal_pipeline,ordinal_col),
    ('cat',cat_pipeline,one_hot_cols)
])
```

```python
x_train,x_test,y_train,y_test=train_test_split(
    x,y,test_size=0.2,random_state=42
)
```

```python
from sklearn.model_selection import cross_val_score
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.svm import SVR
from sklearn.ensemble import RandomForestRegressor
from sklearn.tree import DecisionTreeRegressor
from xgboost import XGBRegressor
```

```python
models={
    "Linear Regression": LinearRegression(),
    "Support Vector Machine": SVR(),
    "Random Forest": RandomForestRegressor(random_state=42),
    "Decision Tree": DecisionTreeRegressor(random_state=42),
    "XGBoost": XGBRegressor(random_state=42)
}
```

```python
for name,model in models.items():
  pipe=Pipeline(steps=[
      ('preprocessor',preprocessor),
      ('model',model)
  ])

  r2_score=cross_val_score(pipe,x_train,y_train,cv=5,scoring='r2')

  print(f"{name} R2 Score: {np.mean(r2_score)}")
```

```python
from sklearn.model_selection import RandomizedSearchCV

# Define the parameter distribution for XGBoost
# We use the 'model__' prefix to refer to the 'model' step in the pipeline
param_dist = {
    'model__n_estimators': [100, 300, 500, 800],
    'model__learning_rate': [0.01, 0.05, 0.1, 0.2],
    'model__max_depth': [3, 5, 7, 9],
    'model__subsample': [0.6, 0.8, 1.0],
    'model__colsample_bytree': [0.6, 0.8, 1.0]
}

# Create the pipeline with the specific XGBoost model
xgb_pipe = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('model', XGBRegressor(random_state=42))
])

# Initialize RandomizedSearchCV
xgb_random_search = RandomizedSearchCV(
    xgb_pipe,
    param_distributions=param_dist,
    n_iter=10,
    cv=5,
    scoring='r2',
    n_jobs=-1,
    random_state=42,
    verbose=1
)

# Fit the search
xgb_random_search.fit(x_train, y_train)

print(f"Best R2 Score: {xgb_random_search.best_score_}")
print(f"Best Parameters: {xgb_random_search.best_params_}")
```

```python
from sklearn.metrics import r2_score as r2_func, mean_absolute_error

# Use the best estimator found by RandomizedSearchCV to make predictions on the test set
best_xgb_model = xgb_random_search.best_estimator_
y_pred = best_xgb_model.predict(x_test)

# Calculate final metrics using the re-imported function
final_r2 = r2_func(y_test, y_pred)
final_mae = mean_absolute_error(y_test, y_pred)

print(f"Final Model Performance on Test Set:")
print(f"R2 Score: {final_r2:.4f}")
print(f"Mean Absolute Error: ${final_mae:.2f}")

# Visualize Predictions vs Actual
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(12, 8))
sns.scatterplot(x=y_test, y=y_pred, alpha=0.5)
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], '--r', linewidth=2)
plt.xlabel('Actual Price')
plt.ylabel('Predicted Price')
plt.title('Actual vs Predicted Diamond Prices (Tuned XGBoost)')
plt.legend()
plt.show()
```
